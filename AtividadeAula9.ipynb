```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
navegador = gs.Chrome()
```

```python
url = (
"https://www.climatempo.com.br/previsao-do-tempo/15-dias/cidade/1403/nao-me-toque-rs"
)
```

```python
navegador.get(url)
sleep(2)
print("Página:", navegador.title)
```

```python
navegador.get(url)
max_temp = navegador.find_element(
  By.XPATH,
  "/html/body/div[4]/div[5]/div[3]/div[1]/div[2]/div[6]/section[1]/div[2]/div/div[3]/div[1]/div[2]/span[2]").text

min_temp = navegador.find_element(
  By.XPATH,
  "/html/body/div[4]/div[5]/div[3]/div[1]/div[2]/div[6]/section[1]/div[2]/div/div[3]/div[1]/div[2]/span[1]").text

print("Cidade: Não-me-Toque")
print(f"Temperatura Mínima: {min_temp}")
print(f"Temperatura Máxima: {max_temp}")
```
